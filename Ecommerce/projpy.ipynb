```python
df.drop(columns=['Unnamed: 22'], inplace=True)
```

```python
print(df['ship-state'].value_counts())
```

```python
df['ship-state'] = df['ship-state'].str.upper().str.strip()
```

```python
print(df['ship-state'].value_counts())
print(df['ship-state'].nunique())
```

```python
state_corrections = {
    'RAJSHTHAN': 'RAJASTHAN',
    'RAJSTHAN': 'RAJASTHAN',
    'ORISSA': 'ODISHA',
    'PB': 'PUNJAB',
    'RJ': 'RAJASTHAN',
    'NL': 'NAGALAND',
    'AR': 'ARUNACHAL PRADESH',
    'NEW DELHI': 'DELHI',
}

df['ship-state'] = df['ship-state'].replace(state_corrections)

# نمسح الصفوف اللي فيها قيم مش ولايات حقيقية أصلاً
df = df[~df['ship-state'].isin(['APO', 'PUNJAB/MOHALI/ZIRAKPUR'])]

print(df['ship-state'].nunique())
```

```python
print(df['Amount'].describe())
print(df['Amount'].isna().sum())
```

```python
print(df[df['Amount'].isna()]['Status'].value_counts())
```

```python
df = df.dropna(subset=['Amount'])
print(df.shape)
```

```python
df['Date']=pd.to_datetime(df['Date'],errors='coerce')
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['ship-city', 'ship-state', 'ship-postal-code', 'ship-country'])
print(df.shape)
```

```python
print(df['Amount'].sum())
```

```python
print(df['Order ID'].nunique())   # Total Orders (unique, not row count)
print(df['Qty'].sum())
```

```python
total_ren=df['Amount'].sum()
AOV=total_ren/113003
print(AOV)
```

```python
group1=df.groupby('Category')['Amount'].sum()
print(group1.sort_values(ascending=False))
```

```python
group2 = df.groupby('Category')['Qty'].sum()
print(group2.sort_values(ascending=False))
```

```python
group2 = df.groupby('Category')['Qty'].sum()
print(group2.sort_values(ascending=False))
```

```python
cancelled = df[df['Status'] == 'Cancelled'].shape[0]
total = df.shape[0]
cancellation_rate = (cancelled / total) * 100
print(f"Cancellation Rate: {cancellation_rate:.2f}%")
```

```python
top_states = df.groupby('ship-state')['Amount'].sum().sort_values(ascending=False)
print(top_states.head(5))
```

```python
import pyodbc

conn = pyodbc.connect(
    'DRIVER={ODBC Driver 17 for SQL Server};'
    'SERVER=.;'
    'DATABASE=ecommerce_dashboard_v2;'
    'Trusted_Connection=yes;'
)

cursor = conn.cursor()
print("Connected successfully!")
```

```python
products_df = df[['SKU', 'Style', 'Category', 'Size', 'ASIN']].drop_duplicates(subset=['SKU'])
print(products_df.shape)
```

```python
for index, row in products_df.iterrows():
    cursor.execute(
        "INSERT INTO products (sku, style, category, size, asin) VALUES (?, ?, ?, ?, ?)",
        row['SKU'], row['Style'], row['Category'], row['Size'], row['ASIN']
    )

conn.commit()
print("Products inserted successfully!")
```

```python
orders_df = df[['Order ID', 'Date', 'Status', 'Fulfilment', 'ship-city', 'ship-state', 'ship-postal-code', 'B2B']].drop_duplicates(subset=['Order ID'])
print(orders_df.shape)
```

```python
for index, row in orders_df.iterrows():
    cursor.execute(
        "INSERT INTO orders (order_id, order_date, status_, fulfilment, ship_city, ship_state, ship_postal_code, b2b) VALUES (?, ?, ?, ?, ?, ?, ?, ?)",
        row['Order ID'], row['Date'], row['Status'], row['Fulfilment'], row['ship-city'], row['ship-state'], row['ship-postal-code'], row['B2B']
    )

conn.commit()
print("Orders inserted successfully!")
```

```python
order_items_df = df[['Order ID', 'SKU', 'Qty', 'Amount']]
print(order_items_df.shape)
```

```python
for index, row in order_items_df.iterrows():
    cursor.execute(
        "INSERT INTO order_items (order_id, sku, qty, amount) VALUES (?, ?, ?, ?)",
        row['Order ID'], row['SKU'], row['Qty'], row['Amount']
    )

conn.commit()
print("Order items inserted successfully!")
```
